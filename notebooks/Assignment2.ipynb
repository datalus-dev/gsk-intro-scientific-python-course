## 1. NumPy Arrays

```python
array = np.arange(6)
walk = array.reshape(2,3)
print(walk)
```

```python
print(walk.ndim)
```

```python
print(walk.shape)
```

```python
print(walk.dtype)
```

```python
zeroax = walk.sum(axis = 0)
print(zeroax)
```

```python
oneax = walk.sum(axis = 1)
print(oneax)
```

When using an axis of zero, Python computes along the y-axis of the array while preserving the number of columns. It adds the columns together (0 + 3 =3, 1+4 =4, 2+5=7) and returns the sum. When using an axis of zero, Python computes along the x-axis while preserving the number of rows. It adds the rows together (0 + 1 + 2 =3 and 3+4+5 =12).

```python
array1d = np.arange(3)
```

```python
broadcast = walk * array1d
print(broadcast)
```

## 2. A Simulation

```python
rng = np.random.default_rng(seed = 3)
```

```python
steps = rng.choice([-1,1],size =(5,500))
```

```python
position = np.cumsum(steps, axis = 1)
print(position)
```

```python
print(position.shape)
```

The shape is 5 rows with 500 columns. The five rows each represent a walk and the 500 columns represent the position that you're at after every step of the walk (after summing), or the direction that you're moving (either 1 or -1).

## 3. Building a Plot in Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
position[0]
```

```python
x = np.arange(len(position[0]))

fig, ax = plt.subplots()
ax.plot(x,position[0], label = 'Walk 1')
ax.plot(x,position[1],label = 'Walk 2' )
ax.plot(x,position[2], label = 'Walk 3')
ax.plot(x,position[3], label = 'Walk 4')
ax.plot(x,position[4], label = 'Walk 5')
ax.set_title('5 Random Walks')
ax.set_xlabel('Step')
ax.set_ylabel('Position')
ax.legend()



```
