```python
arr = np.array([1, 2, 3, 4])
arr.ndim, arr.shape, arr.dtype
```

## Random Numbers and Simulations

`rng.random()` draws a float uniformly between 0 and 1. Passing a seed to `default_rng()` makes your random draws reproducible.

```python
rng = np.random.default_rng()
rng.random()
```

```python
rng = np.random.default_rng(seed=42)
rng.random()
```

### Other Ways to Draw Random Values

```python
rng = np.random.default_rng(seed=42)

rng.integers(1, 7, size=10)          # 10 simulated die rolls
```

```python
rng.choice(["heads", "tails"], size=10)  # 10 simulated coin flips
```

```python
rng.normal(loc=0, scale=1, size=10)  # 10 draws from a standard Normal
```

### Simple Simulations

`rolls == 6` gives an array of `True`/`False`. `True` counts as 1 and `False` as 0, so the mean is the proportion of rolls that were a 6.

```python
rng = np.random.default_rng(seed=42)

rolls = rng.integers(1, 7, size=10_000)
(rolls == 6).mean()   # estimated probability of rolling a 6
```

## Array manipulation

`.reshape()` changes the shape of an array. `.ravel()` flattens it into a 1-d array.

```python
arr = np.arange(24)
arr.reshape(4, 6)
```

```python
arr.reshape(2, 4, 3)
```

```python
arr.reshape(4, 6).ravel()
```

`axis=0` sums down the rows, giving one result per column. `axis=1` sums across the columns, giving one result per row.

```python
arr = np.arange(24).reshape(4, 6)
arr.sum(axis=0).shape   # (6,): one sum per column
```

```python
arr.sum(axis=1).shape   # (4,): one sum per row
```

### Broadcasting

Multiplying arrays with `*` is element-wise; `@` is matrix multiplication. Broadcasting works between two arrays if their trailing dimensions are equal, or one of them is 1.

```python
arr = np.arange(24).reshape([8, 3])
scaler = np.array([1, 2, 3])
new_arr = arr * scaler
new_arr
```

### Combining arrays

`np.hstack()` stacks arrays side by side, adding columns. `np.vstack()` stacks them on top of each other, adding rows. Both take a single tuple or list of arrays.

```python
a = np.arange(6).reshape(2, 3)
np.hstack((a, a)).shape   # (2, 6)
```

```python
np.vstack((a, a)).shape   # (4, 3)
```
